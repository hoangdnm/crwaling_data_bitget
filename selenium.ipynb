```python
import selenium 
from selenium.webdriver.common.by import By
from selenium import webdriver
```

```python
driver=webdriver.Chrome()
url='https://www.bitget.com/price/bitcoin'
driver.get(url)
```

```python
a=driver.find_element(By.XPATH,'//div[@class="grid gap-x-[80px] gap-y-[8px] grid-cols-2 mt-[40px] ltIpad:grid-cols-1 ltIpad:gap-x-0 ltIpad:[mt-24px]"]')
b=a.find_elements(By.TAG_NAME,'span')
```

```python
for i,n in enumerate(b):
    print(i,n.get_attribute('textContent'))
```
